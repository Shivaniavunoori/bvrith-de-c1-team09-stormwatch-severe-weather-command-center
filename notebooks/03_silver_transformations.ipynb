**Week:** 5

**Project:** ShipTrack: Supply Chain Visibility Hub

Goal: Standardize raw Bronze data into clean Silver tables.


```python
# Table 1 - Scan Events
bronze_df = spark.table("bronze_scan_events")

silver_df = bronze_df.dropDuplicates()

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_scan_events")

print("Silver table created: silver_scan_events")
print("Silver row count:", spark.table("silver_scan_events").count())


# Table 2 - Shipments
bronze_df = spark.table("bronze_shipments")

silver_df = bronze_df.dropDuplicates()

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_shipments")

print("Silver table created: silver_shipments")
print("Silver row count:", spark.table("silver_shipments").count())


# Table 3 - Carriers
bronze_df = spark.table("bronze_carriers")

silver_df = bronze_df.dropDuplicates()

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_carriers")

print("Silver table created: silver_carriers")
print("Silver row count:", spark.table("silver_carriers").count())


# Table 4 - Routes
bronze_df = spark.table("bronze_routes")

silver_df = bronze_df.dropDuplicates()

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_routes")

print("Silver table created: silver_routes")
print("Silver row count:", spark.table("silver_routes").count())


# Table 5 - Hubs
bronze_df = spark.table("bronze_hubs")

silver_df = bronze_df.dropDuplicates()

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_hubs")

print("Silver table created: silver_hubs")
print("Silver row count:", spark.table("silver_hubs").count())


# Table 6 - Exceptions
bronze_df = spark.table("bronze_exceptions")

silver_df = bronze_df.dropDuplicates()

display(silver_df.limit(10))

silver_df.write.mode("overwrite").saveAsTable("silver_exceptions")

print("Silver table created: silver_exceptions")
print("Silver row count:", spark.table("silver_exceptions").count())
```

```python
%sql
SELECT 'silver_scan_events' AS table_name, COUNT(*) AS record_count
FROM silver_scan_events

UNION ALL

SELECT 'silver_shipments', COUNT(*)
FROM silver_shipments

UNION ALL

SELECT 'silver_carriers', COUNT(*)
FROM silver_carriers

UNION ALL

SELECT 'silver_routes', COUNT(*)
FROM silver_routes

UNION ALL

SELECT 'silver_hubs', COUNT(*)
FROM silver_hubs

UNION ALL

SELECT 'silver_exceptions', COUNT(*)
FROM silver_exceptions

ORDER BY record_count DESC
```
